```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import fsolve
```

```python
data = pd.read_csv("aol_data.xlsx - in.csv")
data
```

```python
waktu = np.arange(data.shape[1])
produksi = data.iloc[0].values; produksi
```

```python
degree = 3
koefisien = np.polyfit(waktu, produksi, degree)
model = np.poly1d(koefisien)
print('The Polynomial Model is: ')
print(model)

fitwaktu = np.linspace(waktu[0], waktu[-1], 100)
fitproduksi = model(fitwaktu)
```

```python
plt.figure(figsize=(9, 8))
plt.plot(waktu, produksi, '.', label='Jumlah Produksi', color = 'red')
plt.xlabel('Time (in Months)')
plt.ylabel('Production Amount')
plt.title('Production Amount History')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(9, 8))
plt.plot(waktu, produksi, '.', label='Jumlah Produksi', color = 'red')
plt.plot(fitwaktu, fitproduksi, label='Cubic Polynomial Fit', color='yellow')
plt.xlabel('Time (in Months)')
plt.ylabel('Production Amount')
plt.title('Cubic Polynomial Fit to Production Amount History')
plt.legend()
plt.show()
```

```python
max_storage_capacity = 25000
def waktu_max(waktu):
  return model(waktu) - max_storage_capacity

max_month = fsolve(waktu_max, 50)
max_month = np.round(max_month[0], 3)
print(f'Produksi akan melebih kapasitas pada bulan ke-{max_month}')
```

```python
create_starttime = max_month - 13;
print(f'EIGER harus sudah memulai pembangunan pada bulan ke-{create_starttime}')
```
